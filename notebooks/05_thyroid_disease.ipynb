```python
import pandas as pd
import numpy as np

# Load the thyroid dataset
df = pd.read_csv(
    r"../datasets/thyroid.csv"
)

# Dataset overview
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# 1. Check the target distribution
print("Target distribution:")
print(df["Recurred"].value_counts())

print("\nTarget distribution in percentage:")
print(
    (df["Recurred"].value_counts(normalize=True) * 100)
    .round(2)
)

# 2. Inspect the duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())

# 3. Check whether duplicate rows have conflicting targets
duplicate_rows = df[df.duplicated(keep=False)].sort_values(
    by=list(df.columns)
)

display(duplicate_rows.head(10))
```

```python
# Find exact duplicate rows, including the first occurrence
duplicate_mask = df.duplicated(keep=False)
duplicate_df = df[duplicate_mask].sort_values(
    by=list(df.columns)
)

print("Rows involved in duplicate groups:", len(duplicate_df))

# Check whether identical feature values ever have different targets
feature_cols = [col for col in df.columns if col != "Recurred"]

label_counts = (
    df.groupby(feature_cols, dropna=False)["Recurred"]
      .nunique()
)

print(
    "Identical feature groups with conflicting targets:",
    (label_counts > 1).sum()
)
```

```python
# Keep the original dataset unchanged
df_clean = df.drop_duplicates().copy()

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
print("Rows removed:", len(df) - len(df_clean))

print("\nTarget distribution after cleaning:")
print(df_clean["Recurred"].value_counts())

print("\nDuplicate rows remaining:", df_clean.duplicated().sum())
```

```python
# Separate input features and target
X = df_clean.drop(columns=["Recurred"])
y = df_clean["Recurred"]

print("Feature matrix shape:", X.shape)
print("Target vector shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())

print("\nTarget classes:")
print(y.unique())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
# Check how Response relates to the target
print("Response vs Recurred:")
display(pd.crosstab(
    df_clean["Response"],
    df_clean["Recurred"],
    margins=True
))

# Check target distribution for clinical staging features
for col in ["Risk", "T", "N", "M", "Stage"]:
    print(f"\n{col} vs Recurred:")
    display(pd.crosstab(
        df_clean[col],
        df_clean["Recurred"]
    ))
```

```python
# Exclude Response to reduce the risk of target leakage
X = df_clean.drop(columns=["Recurred", "Response"])
y = df_clean["Recurred"]

print("Feature matrix shape:", X.shape)
print("Target vector shape:", y.shape)

print("\nNumerical features:")
print(X.select_dtypes(include="number").columns.tolist())

print("\nCategorical features:")
print(X.select_dtypes(include="object").columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# Identify numerical and categorical columns
numeric_features = X.select_dtypes(include="number").columns.tolist()
categorical_features = X.select_dtypes(include="object").columns.tolist()

# Preprocess numerical features
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Preprocess categorical features
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing steps
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

print("Numerical features:", numeric_features)
print("Number of categorical features:", len(categorical_features))
print("Preprocessing pipeline created successfully.")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),
    "SVM": SVC(
        kernel="rbf",
        probability=True,
        random_state=42
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced"
    )
}

results = []

for name, model in models.items():
    # Combine preprocessing and classifier
    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    # Train the model
    pipeline.fit(X_train, y_train)

    # Predict on unseen test data
    y_pred = pipeline.predict(X_test)

    # Calculate evaluation metrics
    results.append({
        "Model": name,
        "Accuracy": round(accuracy_score(y_test, y_pred) * 100, 2),
        "Precision": round(
            precision_score(y_test, y_pred, pos_label="Yes", zero_division=0) * 100, 2
        ),
        "Recall": round(
            recall_score(y_test, y_pred, pos_label="Yes", zero_division=0) * 100, 2
        ),
        "F1 Score": round(
            f1_score(y_test, y_pred, pos_label="Yes", zero_division=0) * 100, 2
        )
    })

    print(f"\n{'=' * 50}")
    print(name)
    print("=" * 50)
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred, labels=["No", "Yes"]))
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, zero_division=0))

results_df = pd.DataFrame(results)
print("\nMODEL COMPARISON")
print(results_df.to_string(index=False))
```

```python
from sklearn.metrics import accuracy_score
import numpy as np

# Compare predictions from each trained pipeline
predictions = {}

for name, model in models.items():
    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    pipeline.fit(X_train, y_train)
    predictions[name] = pipeline.predict(X_test)

print("Prediction comparison:")

model_names = list(predictions.keys())

for i in range(len(model_names)):
    for j in range(i + 1, len(model_names)):
        first = model_names[i]
        second = model_names[j]

        same = np.array_equal(
            predictions[first],
            predictions[second]
        )

        print(f"{first} vs {second}: Identical = {same}")

print("\nNumber of positive predictions:")
for name, pred in predictions.items():
    print(f"{name}: {(pred == 'Yes').sum()} Yes, {(pred == 'No').sum()} No")
```

```python
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

comparison = []

for name, y_pred in predictions.items():
    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_pred,
        labels=["No", "Yes"]
    ).ravel()

    comparison.append({
        "Model": name,
        "Accuracy (%)": round(accuracy_score(y_test, y_pred) * 100, 2),
        "Recurrence Precision (%)": round(
            precision_score(y_test, y_pred, pos_label="Yes", zero_division=0) * 100, 2
        ),
        "Recurrence Recall (%)": round(
            recall_score(y_test, y_pred, pos_label="Yes", zero_division=0) * 100, 2
        ),
        "Recurrence F1 (%)": round(
            f1_score(y_test, y_pred, pos_label="Yes", zero_division=0) * 100, 2
        ),
        "True Negatives": tn,
        "False Positives": fp,
        "False Negatives": fn,
        "True Positives": tp
    })

comparison_df = pd.DataFrame(comparison)
print(comparison_df.to_string(index=False))
```

```python
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.metrics import make_scorer, precision_score, recall_score, f1_score

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "accuracy": "accuracy",
    "precision_yes": make_scorer(
        precision_score, pos_label="Yes", zero_division=0
    ),
    "recall_yes": make_scorer(
        recall_score, pos_label="Yes", zero_division=0
    ),
    "f1_yes": make_scorer(
        f1_score, pos_label="Yes", zero_division=0
    )
}

cv_results = []

for name, model in models.items():
    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    cv_results.append({
        "Model": name,
        "Accuracy Mean (%)": round(scores["test_accuracy"].mean() * 100, 2),
        "Accuracy Std (%)": round(scores["test_accuracy"].std() * 100, 2),
        "Precision Yes (%)": round(scores["test_precision_yes"].mean() * 100, 2),
        "Recall Yes (%)": round(scores["test_recall_yes"].mean() * 100, 2),
        "F1 Yes (%)": round(scores["test_f1_yes"].mean() * 100, 2)
    })

cv_results_df = pd.DataFrame(cv_results)
print(cv_results_df.to_string(index=False))
```

```python
import joblib
from pathlib import Path
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Build the final pipeline
thyroid_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

# Train using the complete training set
thyroid_model.fit(X_train, y_train)

# Final evaluation on the untouched test set
y_pred_final = thyroid_model.predict(X_test)

print("Final Model: Logistic Regression")
print(f"Test Accuracy: {accuracy_score(y_test, y_pred_final) * 100:.2f}%")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_final, labels=["No", "Yes"]))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_final, zero_division=0))
```
